Medidas de correlación

Primeramente vamos a dar una pequeña definición de que son las medidas de correlación.

La medidas de correlación se aplican a los valores que toman dos variables distintas en cada uno de los elementos de una muestra, y expresan si ambas variables varían a la vez, es decir, si los valores de una son más altos o bajos cuanto más altos o bajos son los de la otra.

 Si los valores de ambas variables aumentan o disminuyen a la vez, su correlación es positiva; en cambio, si los valores de una variable aumentan cuando los de la otra disminuyen, o viceversa, su correlación es negativa.

 Estás son las principales Medidas de correlación:

 * Coeficiente de correlación de Pearson
 * Coeficiente de correlación de Spearman
 * Coeficiente de correlación de Kendall
 * Correlación de Cramér



**Coeficiente de correlación de Pearson**

Esto es una prueba que mide la relación estadistica entre dos variables continuas. Si la asociación entre los elementos no es lineal, entonces el coeficiente no se encuentra representado adecuadamente.

image.png

Este coeficiente posee algunas propiedades.



*   -1 a menos o igual que 1
*   Si el coeficiente de correlación se acerca a 1 **x** aumenta, así mismo **y** también aumenta
*   El valor del coeficiente de correlación no cambia cuando **x**, **y** se multiplican por un número bajo.





**Aplicación con Python, calculo del coeficiente de correlación entre dos columnas númericas**

* Utilizando numpy

```python
import numpy as np

# Datos de ejemplo
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

# Cálculo del coeficiente de correlación de Pearson
correlacion = np.corrcoef(x, y)[0, 1]

print("Coeficiente de correlación de Pearson:", correlacion)
```

* Utilizando scipy

```python
from scipy.stats import pearsonr

# Datos de ejemplo
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Cálculo del coeficiente de correlación de Pearson
correlacion, p_valor = pearsonr(x, y)

print("Coeficiente de correlación de Pearson:", correlacion)
print("Valor p:", p_valor)
```

* Utilizando pandas

```python
import pandas as pd

# Datos de ejemplo
data = {'x': [1, 2, 3, 4, 5], 'y': [2, 4, 6, 8, 10]}
df = pd.DataFrame(data)

# Cálculo del coeficiente de correlación de Pearson
correlacion = df['x'].corr(df['y'])

print("Coeficiente de correlación de Pearson:", correlacion)
```

**Coeficiente de correlación de Spearman**

Es una medida no paramétrica de la correlación de rango. Mide la fuerza y la dirección de la asociación entre dos variables clasificadas, siendo la contrapartida no paramétrica de la correlación de Pearson. Por lo tanto, en este caso no se requiere una distribución normal de los datos.

```python
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import spearmanr

tips = sns.load_dataset("tips")

coef, p = spearmanr(tips['total_bill'], tips['tip'])

print('Correlación Spearman: %.3f' % coef)

alpha = 0.05
if p > alpha:
  print('Las muestras no están correlacionadas p=%.3f' % p)
else:
  print('Las muestras están fuertemente correlacionadas p=%.3f' % p)

print("")

# Crear un gráfico de dispersión
plt.figure(figsize=(10, 6))
sns.scatterplot(x='total_bill', y='tip', data=tips, color='blue')

# Añadir una línea de tendencia basada en una regresión lineal
sns.regplot(x='total_bill', y='tip', data=tips, scatter=False, color='red', line_kws={"label":"Tendencia Lineal"})

# Añadir títulos y etiquetas
plt.title('Relación entre Total de la Factura y la Propina')
plt.xlabel('Total de la Factura (Total Bill)')
plt.ylabel('Propina (Tip)')

# Mostrar la leyenda
plt.legend()

# Mostrar el gráfico
plt.show()
```

**Coeficiente de correlación de Kendall**

 Evalúa la similitud en el ordenamiento de los datos cuando se clasifican por cada una de las variables. Se basa en la idea de pares concordantes y discordantes.
 Un par concordante es un par de observaciones en el cual el orden de las dos variables es el mismo.

Los valores del coeficiente de Kendall pueden oscilar entre -1 y 1. Un valor positivo indica una asociación positiva. Un valor negativo indica asociación negativa. Mientras más alta sea la magnitud, más fuerte será la asociación.

```python
from scipy.stats import kendalltau

tips = sns.load_dataset("tips")
coef, p = kendalltau(tips['total_bill'], tips['tip'])
print("Coeficiente de Kendall: ", coef)
```

**Correlación de Cramér**

Medida del grado de asociación o relación entre dos series de atributos o variables. Se usa únicamente cuando tenemos sólo información categórica acerca de uno o de ambos conjuntos de atributos consiste en una serie no ordenada de categorías.

```python
import pandas as pd
import scipy.stats as stats

tips = sns.load_dataset("tips")
contingency_table = pd.crosstab(tips['sex'], tips['smoker'])

# Calcular la estadística chi-cuadrado, el p-valor, los grados de libertad y las frecuencias esperadas
chi2, p, dof, expected = stats.chi2_contingency(contingency_table)

# Calcular el tamaño de la muestra
n = contingency_table.sum().sum()

# Calcular el coeficiente de Cramér (V de Cramér)
cramers_v = (chi2 / (n * (min(contingency_table.shape) - 1))) ** 0.5

print(f"Coeficiente de correlación de Cramér: {cramers_v:}")
print("")

# Crear el mapa de calor de la tabla de contingencia
plt.figure(figsize=(8, 6))
sns.heatmap(contingency_table, annot=True, cmap='YlGnBu', fmt='d', cbar=True)
plt.title('Tabla de Contingencia entre Sexo y Fumador')
plt.xlabel('Fumador')
plt.ylabel('Sexo')
plt.show()
```

**¿Cuál es la probabilidad de que el número de trabajadores seleccionados del turno de la noche sea igual a 3?**

```python
from scipy.stats import hypergeom

# Número total de trabajadores
Dia = 20
Tarde = 15
Noche = 10
n = 6
probabilidadTres = 3
total = 45

# Crear listas de nombres para cada turno
nombresDia = [f"Dia{i+1}" for i in range(Dia)]
nombresTarde = [f"Tarde{i+1}" for i in range(Tarde)]
nombresNoche = [f"Noche{i+1}" for i in range(Noche)]

probabilidad = hypergeom.pmf(probabilidadTres, total, Noche, n)

print(f"La probabilidad de seleccionar exactamente {probabilidadTres} trabajadores del turno de la noche es: {probabilidad:.4f}")
```

**¿Cuál es la probabilidad de que los empleados seleccionados sean todos del mismo turno?**

```python
from scipy.stats import hypergeom

total = 45
n = 6

# Número total de trabajadores
Dia = 20
Tarde = 15
Noche = 10

# Probabilidad de que todos los seleccionados sean del turno de día
probabilidadDia = hypergeom.pmf(n, total, Dia, Dia)

# Probabilidad de que todos los seleccionados sean del turno de tarde
probabilidadTarde = hypergeom.pmf(n, total, Tarde, Tarde)

# Probabilidad de que todos los seleccionados sean del turno de la noche
probabilidadNoche = hypergeom.pmf(n, total, Noche, Noche)

# Sumar las probabilidades para obtener la probabilidad total
probabilidadTotal = probabilidadDia + probabilidadTarde + probabilidadNoche

print(f"La probabilidad de que todos los seleccionados sean del turno de día es: {probabilidadDia:.4f}")
print(f"La probabilidad de que todos los seleccionados sean del turno de tarde es: {probabilidadTarde:.4f}")
print(f"La probabilidad de que todos los seleccionados sean del turno de la noche es: {probabilidadNoche:.4f}")
print(f"La probabilidad total de que todos los seleccionados pertenezcan al mismo turno es: {probabilidadTotal:.4f}")
```

**Pruebas de normalidad existen para las variables aleatorias**

*   Kolmogorov-Smirnov
*   Shapiro-Wilk
*   Anderson Darling

**Kolmogorov-Smirnov**

Compara la función de distribución acumulada observada de una variable con una distribución teórica determinada, que puede ser la normal, la uniforme, la de Poisson o la exponencial,

Es decir u que se define como una prueba estadística no paramétrica que cuantifica la diferencia entre la función de distribución empírica de una muestra y la función de distribución acumulativa de una distribución de referencia, o las funciones de distribución empíricas de dos muestras.

* **Hipótesis Nula** : Los datos provienen de una distribución normal.
* **Estadísticos** : Compara la distribución acumulativa empírica con la distribución acumulativa de una distribución normal.
* **Interpretación** : Un p-valor bajo sugiere que los datos no siguen una distribución normal.

Para ilustrar la interpretación de datos en python lo haremos a travez de un ejemplo.

**Fórmulas**

image.png

La estadística de prueba es la máxima diferencia absoluta entre la cdf empírica calculada a partir de x y la cdf hipotética: D * = max x ( | F ^ ( x ) − G ( x ) | ) , donde F ^ ( x ) es la cdf empírica y G ( x ) es la cdf de la distribución hipotética.

```python
import pandas as pd
from scipy.stats import kstest
import numpy as np

np.random.seed(0)
df = pd.DataFrame({
    'columna': np.random.normal(loc=0, scale=1, size=100)
})

# Aplicar la prueba de Kolmogorov-Smirnov
stat, p = kstest(df['columna'], 'norm')

# Mostrar los resultados
print(f'Estadístico: {stat}, p-valor: {p}')

if p > 0.05:
    print("Los datos siguen una distribución normal")
else:
    print("Los datos no siguen una distribución normal")

```

* Si el p-valor (p) es mayor que 0.05, se concluye que los datos no difieren significativamente de una distribución normal.
* Si el p-valor es menor o igual a 0.05, se concluye que los datos no siguen una distribución normal.

**Shapiro-Wilk**

La prueba de normalidad de Shapiro-Wilk es aplicable cuando se analizan muestras compuestas por menos de 50 elementos

* Hipótesis Nula (H₀): Los datos provienen de una distribución normal.
* Estadístico: El estadístico de Shapiro-Wilk mide la desviación de los datos respecto a una distribución normal.
* Interpretación: Un p-valor bajo (generalmente < 0.05) indica que se rechaza la hipótesis nula, es decir, los datos no siguen una distribución normal.

**Fórmulas**

image.png

Donde Yi son los datos de la muestra, ordenados por tamaño -ordenado-. Ahora bien, si los datos de la muestra son en realidad una muestra aleatoria de una distribución normal con media desconocida μ y varianza σ2

```python
import pandas as pd
from scipy.stats import shapiro

df = pd.DataFrame({
    'columna': [2.5, 3.6, 2.9, 3.2, 3.0, 3.1, 3.8, 2.7]
})

# Aplicar la prueba de Shapiro-Wilk
stat, p = shapiro(df['columna'])

print(f'Estadístico: {stat}, p-valor: {p}')
if p > 0.05:
    print("Los datos siguen una distribución normal")
else:
    print("Los datos no siguen una distribución normal")
```

**Anderson Darling**

Mide qué tan bien siguen los datos una distribución específica, siendo que, para un conjunto de datos y distribución en particular, mientras mejor se ajuste la distribución a los datos, menor será este estadístico.
Es decir mide el área entre la línea ajustada -basada en la distribución normal- y la función de distribución empírica -que se basa en los puntos de los datos-.

* Hipótesis Nula (H₀): Los datos provienen de una distribución normal.
* Estadístico: Considera tanto la asimetría como la curtosis de los datos para evaluar la normalidad.
* Interpretación: Un p-valor bajo indica que los datos no siguen una distribución normal.
